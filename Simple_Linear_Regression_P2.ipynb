## Step 1: Import necessary libraries

```python
#23070521078, Krishna Wairagade
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

## Step 2: Define the input data (x) and output data (y)

```python
# Input: Hours studied (x)
x = np.array([1, 2, 3, 4, 5])

# Output: Marks obtained (y)
y = np.array([20, 40, 50, 65, 80])

print("Input (Hours):", x)
print("Output (Marks):", y)
```

## Step 3: Reshape the input data for the model

```python
# Scikit-learn expects a 2D array for the input features
X = x.reshape(-1, 1)

print("Reshaped Input (X):\n", X)
```

## Step 4: Create and train the Simple Linear Regression model

```python
# Create a Linear Regression model object
model = LinearRegression()

# Train the model using the input (X) and output (y) data
model.fit(X, y)

print("Model training complete.")
print(f"Coefficient (slope): {model.coef_[0]:.2f}")
print(f"Intercept: {model.intercept_:.2f}")
```

## Step 5: Make predictions using the trained model

```python
# Predict y values using the trained model
y_pred = model.predict(X)

print("Actual Y values:", y)
print("Predicted Y values:", np.round(y_pred, 2))
```

## Step 6: Calculate evaluation metrics (MAE, MSE, RMSE)

```python
# Calculate Mean Absolute Error (MAE)
mae = mean_absolute_error(y, y_pred)

# Calculate Mean Squared Error (MSE)
mse = mean_squared_error(y, y_pred)

# Calculate Root Mean Squared Error (RMSE)
rmse = np.sqrt(mse)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
```
